# 02 — CKD Exploratory Data Analysis
**Objective:** Understand the CKD dataset structure, distributions, missing values, and class balance.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../ml/data/raw/kidney_disease.csv')
df.columns = df.columns.str.strip()
print("Shape:", df.shape)
df.head()
```

## Missing Values

```
missing = df.isnull().sum().sort_values(ascending=False)
print(missing[missing > 0])
plt.figure(figsize=(12,4))
missing[missing > 0].plot(kind='bar', color='coral')
plt.title('Missing Values per Column — CKD Dataset')
plt.tight_layout()
plt.show()
```

## Class Distribution

```
if 'classification' in df.columns:
    df['classification'] = df['classification'].str.strip()
    print(df['classification'].value_counts())
    df['classification'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(5,5),
                                              colors=['#22c55e','#ef4444'])
    plt.title('CKD Class Distribution')
    plt.show()
```

## Numeric Feature Distributions

```
numeric = df.select_dtypes(include='number').columns.tolist()
df[numeric].hist(figsize=(16,10), bins=20, color='steelblue', edgecolor='white')
plt.suptitle('CKD Numeric Feature Distributions')
plt.tight_layout()
plt.show()
```

## Correlation Heatmap

```
plt.figure(figsize=(12,8))
sns.heatmap(df[numeric].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('CKD Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```
