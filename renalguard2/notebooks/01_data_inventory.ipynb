# 01 — Data Inventory
**Objective:** Verify dataset files, inspect shapes, column names, dtypes, and class distributions.

```
import pandas as pd
from pathlib import Path

DATA = Path('../ml/data/raw')
datasets = {
    'CKD':          'kidney_disease.csv',
    'DIABETES':     'diabetes_prediction.csv',
    'HYPERTENSION': 'hypertension.csv',
}
for name, fname in datasets.items():
    path = DATA / fname
    if path.exists():
        df = pd.read_csv(path)
        print(f"{'='*50}")
        print(f"Dataset : {name}")
        print(f"File    : {fname}")
        print(f"Shape   : {df.shape}")
        print(f"Columns : {df.columns.tolist()}")
        print(f"Dtypes  : {df.dtypes.value_counts().to_dict()}")
        print(f"Nulls   : {df.isnull().sum().sum()}")
    else:
        print(f"⚠️  Missing: {path}")
```

## Class Distribution

```
for name, fname in datasets.items():
    path = DATA / fname
    if not path.exists(): continue
    df = pd.read_csv(path)
    # Find target column
    for col in ['classification','diabetes','hypertension','target','label','output']:
        if col in [c.lower() for c in df.columns]:
            target_col = [c for c in df.columns if c.lower() == col][0]
            print(f"{name} — {target_col}: {df[target_col].value_counts().to_dict()}")
            break
```
